# Explore saved out-of-sample results
Run the production data and training commands first. This notebook only visualizes saved outputs. Daily reference rates are not executable quotes.

```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
results = json.loads((root / 'models/run/evaluation.json').read_text())
pd.DataFrame([{'fold': f['fold'], 'regime_rmse': f['regime']['rmse'], 'baseline_rmse': f['global_xgboost']['rmse']} for f in results['folds']]).plot.bar(x='fold', title='Chronological holdout RMSE')
```
